#Step 1: Calibrate the risk models

This step is not needed for the purposes of reproducing results but the steps that were taken to run the calibrariton are outlined here. These steps may need to be re-run should new diseases be added to mSHIFT, the risk models updated or additional disease risk associations are included for more food groups.  

This step must be run three times for each each outcome: diabetes, CVD and mortalities.

Before running, navigate to the disease module in question and the corresponding risk prediction function:

* diabetes.py: Diabetes_risk_Alva()
* CVD.py: updated_Framingham_CVD_risk()
* mortalities.py: mortality_prob()

and find the calibration_ within the function. This is either a dictionary of a list of tuples or a list of tuples depending on whether the calibrariton is by age and sex or just by age. The tuples are of the form (age_min, age_max, calibrration_constant). The observed incidence for each health outcome with which to compare the predicted incidence are set within the demographic_conditions.py module. Prior to calibration the calibrartion values should be set to 1 for all demographic groups.

For example, wihin the diabetes.py module, prior to calibrartion the values should look like:

```
    age_calibration_values = [
        (15, 19, 1),
        (20, 29, 1),
        (30, 39, 1),
        (40, 49, 1),
        (50, 59,  1),
        (60, 69, 1),
        (70, 200, 1)
    ]

```

After running risk_calibrariton.py the value for the calibration constant will appear in the terminal for each age group. These values should then be inserted into the disease risk function for the corresponding age group. The final calibration values for the diabetes risk model are given by:


```
    age_calibration_values = [
        (15, 19, 82.51),
        (20, 29, 17.46),
        (30, 39, 5.73),
        (40, 49, 1.74),
        (50, 59,  1.78),
        (60, 69, 1.75),
        (70, 200, 1.17)
    ]

```

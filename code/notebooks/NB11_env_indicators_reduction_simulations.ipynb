This notebook calculates the impact of each meat and dairy reduction scenario on the additional indicators from foodDB.

```python
import pandas as pd
import numpy as np
import json
import os
from pathlib import Path
import sys

data_path = Path("data")
results_path = Path("results")
```

```python
# Add the parent directory of the notebook to sys.path, enables module imports from notebook_code
sys.path.append(str(Path().resolve() / 'code/notebooks/notebook_code'))
```

```python
from reduction_simulation_env_indicators import new_diet_data, apply_reductions_env_data, include_env_data_ccc, include_env_data_high_consumers
from data_processing import nutrient_intake
```

```python
meat_food_groups = np.loadtxt(data_path / 'indicator_lists/meat_food_groups.txt', dtype = str).tolist()
all_meat_food_groups = np.loadtxt(data_path / 'indicator_lists/all_meat_food_groups.txt', dtype = str).tolist()
food_groups_dairy = np.loadtxt(data_path / 'indicator_lists/food_groups_dairy.txt', dtype = str).tolist()
env_columns = np.loadtxt(data_path / 'indicator_lists/env_columns.txt', dtype = str).tolist()
error_columns = np.loadtxt(data_path / 'indicator_lists/error_columns.txt', dtype = str).tolist()
food_groups_dairy = np.loadtxt(data_path / 'indicator_lists/food_groups_dairy.txt', dtype = str).tolist()
```

```python
nutrients_per_gram_meat = pd.read_parquet(data_path / 'nutrients_per_gram_meat.parquet')
nutrients_per_gram_dairy = pd.read_parquet(data_path / 'nutrients_per_gram_dairy.parquet')
```

```python
# Should now inlcude environmental indicators and dairy disaggregation
diet_data = pd.read_parquet(data_path / 'diet_data.parquet')
```

```python
# Dictionary mapping composite food items containing each meat food type to the associated non-composite meat ingredient in that food type
with open(data_path / 'mappings/meat_ingredient_map_desc.json', 'r') as f:
    meat_mapping_dict = json.load(f)

# Dictionary mapping composite food items containing each dairy food type to the associated dairy ingredient item in that food type
with open(data_path / 'mappings/dairy_ingredients_dict.json', 'r') as f:
    dairy_mapping_dict = json.load(f)

# Dictionary mapping composite dairy food codes
with open(data_path / 'mappings/dairy_ingredient_proportion_dict.json', 'r') as f:
    dairy_ingredient_proportion_dict = json.load(f)
```

Run the scenarios for the CCC percentage reductions in meat for all consumers.

```python
for meat_reduction in [0, 20, 35]:
  scenario_path =  results_path / f'Output/Percent_reduction/{meat_reduction}.0_reduction20.0_dairy_reduction/nutrient_intake/'
  include_env_data_ccc(
                      diet_data_baseline = diet_data,
                      env_columns= env_columns,
                      error_columns = error_columns,
                      food_groups_dairy = food_groups_dairy,
                      nutrients_per_gram_meat = nutrients_per_gram_meat,
                      nutrients_per_gram_dairy = nutrients_per_gram_dairy,
                      meat_mapping_dict = meat_mapping_dict,
                      dairy_mapping_dict = dairy_mapping_dict,
                      dairy_ingredient_proportion_dict = dairy_ingredient_proportion_dict,
                      scenario_path = scenario_path
                      )
```

Save the individual level data to a path with the corresponding scenario name

```python
# Scenario label in directory to the percent reduction in all meat consumption
ccc_scenario_mapping = {20: 22,
                        35: 23,
                        0: 34}

# Path where results for all scenarios, including substitution scenarios will be stored
base_path = results_path / 'Output/Scenarios/'
base_path.mkdir(parents=True, exist_ok=True)

# Copy the ccc reduction scenarios into a directory with the assocaited scenario label
for meat_reduction in [0, 20, 35]:
  original_scenario_path =  results_path / f'Output/Percent_reduction/{meat_reduction}.0_reduction20.0_dairy_reduction/nutrient_intake/'
  df_scenario = pd.read_parquet(original_scenario_path / 'df_scenario.parquet')
  final_scenario_path = base_path / f"Scenario_{ccc_scenario_mapping[meat_reduction]}/"
  final_scenario_path.mkdir(parents=True, exist_ok=True)
  df_scenario.to_parquet(final_scenario_path / f'df_scenario{ccc_scenario_mapping[meat_reduction]}.parquet')
```

Apply the reductions to each of the scenarios meeting the CCC targets through reductions in red meat among high consumers

```python
# Submitted results correspond to setting max_seed = 51.
max_seed = 51

for max_intake in [70, 60, 31]:
  path_scenario = results_path / f"Output/Max_intake/{max_intake}.0_max_reduction20.0_dairy_reduction/nutrient_intake/"
  include_env_data_high_consumers(diet_data_baseline = diet_data,
                                  max_intake = max_intake,
                                  max_seed = max_seed,
                                  env_columns = env_columns,
                                  error_columns = error_columns,
                                  food_groups_dairy = food_groups_dairy,
                                  nutrients_per_gram_meat = nutrients_per_gram_meat,
                                  nutrients_per_gram_dairy = nutrients_per_gram_dairy,
                                  meat_mapping_dict = meat_mapping_dict,
                                  dairy_mapping_dict = dairy_mapping_dict,
                                  dairy_ingredient_proportion_dict = dairy_ingredient_proportion_dict,
                                  path_scenario = path_scenario
                                  )
```
